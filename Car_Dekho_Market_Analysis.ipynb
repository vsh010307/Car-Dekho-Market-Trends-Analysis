```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("P3-Car Market Trends Analysis with Car Dekho Data.csv")

# Remove duplicates
df = df.drop_duplicates()

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())


# 1. FUEL TYPE


plt.figure(figsize=(7,5))
sns.countplot(data=df, x="Fuel_Type")
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.show()



# 2. TRANSMISSION


plt.figure(figsize=(7,5))
sns.countplot(data=df, x="Transmission")
plt.title("Number of Cars by Transmission Type")
plt.xlabel("Transmission")
plt.ylabel("Number of Cars")
plt.show()



# 3. PRESENT PRICE VS SELLING PRICE


correlation = df["Present_Price"].corr(df["Selling_Price"])

print("Present Price vs Selling Price Correlation:",
      round(correlation, 3))

plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x="Present_Price",
    y="Selling_Price"
)
plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price (₹ lakh)")
plt.ylabel("Selling Price (₹ lakh)")
plt.show()


# -------------------------------
# 4. AVERAGE SELLING PRICE
# -------------------------------

avg_price = df.groupby("Fuel_Type")["Selling_Price"].mean()

print("\nAverage Selling Price by Fuel Type:")
print(avg_price)

plt.figure(figsize=(7,5))
avg_price.plot(kind="bar")
plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price (₹ lakh)")
plt.xticks(rotation=0)
plt.show()



# FINAL RESULTS


print("\n========== FINAL RESULTS ==========")

print("Total Records:", len(df))
print("Average Selling Price:",
      round(df["Selling_Price"].mean(), 2), "lakh")
print("Average Present Price:",
      round(df["Present_Price"].mean(), 2), "lakh")
print("Average Kms Driven:",
      round(df["Kms_Driven"].mean(), 2), "km")

print("\nFuel Type:")
print(df["Fuel_Type"].value_counts())

print("\nTransmission:")
print(df["Transmission"].value_counts())

print("\nSeller Type:")
print(df["Seller_Type"].value_counts())
```
